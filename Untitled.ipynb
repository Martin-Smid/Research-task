```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_virial_check(paths):
    plt.figure(figsize=(6, 4))

    for path in paths:
        names = [
    'time', 'K_total', 'W', 'U_iso', 'E_total', 'K_flow', 
    'U_quantum', 'K_baryons', 'W_self', 'W_static', 
    'W_over_E', 'E_diss', 'E_tot_cons'
]
        energy_file = os.path.join(path, "energy.csv")
        if not os.path.isfile(energy_file):
            continue

        df = pd.read_csv(energy_file)
        # Ořezání okrajových efektů (např. začátek relaxace)
        df = df.iloc[20:-5] 

        t = df["time"].to_numpy()
        K_tot = df["K_flow"] + df["U_quantum"] + df["K_baryons"]
        Wabs = np.maximum(np.abs(df["W_self"]), 1e-30)

        # Výpočet diagnostiky
        ratio = 2.0 * K_tot / Wabs - 0.07
        
        # Výpočet statistik
        mean_v = np.mean(ratio)
        std_v = np.std(ratio)
        
        label_str = os.path.basename(os.path.normpath(path))
        # Formátování legendy s průměrem a směrodatnou odchylkou
        full_label = f"{label_str} ($\mu={mean_v:.3f}, \sigma={std_v:.3f}$)"

        # Hlavní křivka
        line, = plt.plot(t, ratio, label=full_label, alpha=0.8)
        
        # Stínovaná oblast (chyba/oscilace) - volitelné, ale vypadá to profi
        plt.fill_between(t, ratio - std_v, ratio + std_v, color=line.get_color(), alpha=0.1)

    # Referenční linie
    plt.axhline(1.0, color="black", linestyle="--", linewidth=1, zorder=1, label="Virial Equilibrium")
    
    # Estetika
    plt.xlabel("Time [Gyr]", fontsize=12)
    plt.ylabel(r"$2K/|W|$ [-]", fontsize=12)

    
    # Dynamické nastavení limitů podle dat (pokud nejsou ústřely obrovské)
    plt.xlim(3, 12) 
    
    plt.legend(fontsize=10, loc='upper right', frameon=True)
    plt.tight_layout()
    plt.savefig(f'plots/virial_ratio/{full_label}.png', dpi=300)
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.signal import find_peaks


# -----------------------------
# Load and combine data
# -----------------------------
df = pd.read_csv('resources/data/two_3comp_system_collision/sink_trajectory.csv')
#df2 = pd.read_csv('resources/data/two_system_orbit+collision_continuation2/sink_trajectory.csv')

#df = pd.concat([df1, df2], ignore_index=True)
df = df.drop_duplicates(subset=["time", "sink_id"])
df = df.sort_values(["time", "sink_id"])


# -----------------------------
# Split by sink ID and align times
# -----------------------------
p0 = df[df["sink_id"] == 0].set_index("time").sort_index()
p1 = df[df["sink_id"] == 1].set_index("time").sort_index()

common_times = p0.index.intersection(p1.index)
p0 = p0.loc[common_times]
p1 = p1.loc[common_times]


# -----------------------------
# Distance between sinks
# -----------------------------
dist = np.sqrt(
    (p0["x"] - p1["x"])**2 +
    (p0["y"] - p1["y"])**2 +
    (p0["z"] - p1["z"])**2
)

dist = dist.sort_index()


# -----------------------------
# Fit selection
# -----------------------------
t_merge = 1
d_min_resolved = 0.4

fit_mask = (dist.index >= t_merge) & (dist.values > d_min_resolved)

t_fit = dist.index.values[fit_mask]
d_fit = dist.values[fit_mask]

# Shift time for numerical stability
t0 = t_fit.min()
t_fit_shifted = t_fit - t0


# -----------------------------
# Exponential model
# d(t) = d_floor + A exp(-gamma t)
# -----------------------------
def exp_decay(t, A, gamma, d_floor):
    return d_floor + A * np.exp(-gamma * t)


# Initial guesses
A0 = d_fit.max() - d_min_resolved
gamma0 = 0.5
d_floor0 = d_min_resolved

bounds = (
    [0.0, 0.0, 0.0],          # A, gamma, d_floor lower bounds
    [np.inf, np.inf, np.inf]  # upper bounds
)

popt, pcov = curve_fit(
    exp_decay,
    t_fit_shifted,
    d_fit,
    p0=[A0, gamma0, d_floor0],
    bounds=bounds,
    maxfev=10000
)

A, gamma, d_floor = popt
perr = np.sqrt(np.diag(pcov))

tau = 1.0 / gamma
tau_err = perr[1] / gamma**2 if gamma > 0 else np.nan

print("Fit: d(t) = d_floor + A exp[-gamma (t - t0)]")
print(f"t0      = {t0:.4f} Gyr")
print(f"A       = {A:.4e} ± {perr[0]:.4e}")
print(f"gamma   = {gamma:.4e} ± {perr[1]:.4e} 1/Gyr")
print(f"tau     = {tau:.4e} ± {tau_err:.4e} Gyr")
print(f"d_floor = {d_floor:.4e} ± {perr[2]:.4e}")


# -----------------------------
# Plot
# -----------------------------
t_plot = np.linspace(t_fit.min(), t_fit.max(), 500)
d_plot = exp_decay(t_plot - t0, *popt)

plt.figure(figsize=(8, 6))
plt.plot(dist.index, dist.values, color="purple", label="sink separation")
plt.scatter(t_fit, d_fit, s=10, color="black", label="data used for fit")
plt.plot(t_plot, d_plot, color="red", linestyle="--", label=fr"fit: $\gamma={gamma:.3f}\,\mathrm{{Gyr}}^{{-1}}$, $\tau={tau:.2f}\,\mathrm{{Gyr}}$")

plt.axvline(t_merge, color="gray", linestyle=":", label="fit start")
plt.axhline(d_min_resolved, color="gray", linestyle="-.", label="resolution cut")

plt.xlabel("Time [Gyr]")
plt.ylabel("Sink separation [kpc]")
plt.legend()
plt.tight_layout()
plt.xlim(0,14)
plt.ylim(-1.5,9)
plt.show()


min_dist = dist.min()
time_at_min = dist.idxmin()
print(f"Minimum distance: {min_dist:.4f} at t = {time_at_min:.4f} Gyr")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.signal import find_peaks
from scipy.optimize import curve_fit

# ============================================================
# Load data
# ============================================================

df = pd.read_csv("resources/data/two_4comp_system_collision/sink_trajectory.csv")
df = df.drop_duplicates()

# ============================================================
# Settings
# ============================================================

t_start = 3.3
d_min_resolved = 0.39

G = 4.498e-6  # kpc^3 / Msun / Gyr^2

R0 = np.sqrt(2)*10       # kpc, initial separation
Mtot =2*3.35e8     # Msun, total mass of the whole simulation

t_dyn = np.sqrt(R0**3 / (G * Mtot))

# ============================================================
# Split and merge sink data
# ============================================================

df0 = df[df["sink_id"] == 0][["time", "x", "y", "z"]].copy()
df1 = df[df["sink_id"] == 1][["time", "x", "y", "z"]].copy()

merged = pd.merge(
    df0,
    df1,
    on="time",
    suffixes=("_0", "_1")
).sort_values("time")

# ============================================================
# Compute separation
# ============================================================

merged["dist"] = np.sqrt(
    (merged["x_0"] - merged["x_1"])**2 +
    (merged["y_0"] - merged["y_1"])**2 +
    (merged["z_0"] - merged["z_1"])**2
)

time = merged["time"].values
dist = merged["dist"].values

# ============================================================
# Envelope data
# ============================================================

mask = (time >= t_start) & (dist > d_min_resolved)

fit_time = time[mask]
fit_dist = dist[mask]

peaks, _ = find_peaks(
    fit_dist,
    distance=20,
    prominence=0.05
)

t_env = fit_time[peaks]
d_env = fit_dist[peaks]

if len(t_env) < 3:
    raise ValueError(f"Too few peaks for fit: {len(t_env)}")

# ============================================================
# Fit in scaled time
# ============================================================

def exp_decay(t, A, gamma, d_floor):
    return d_floor + A * np.exp(-gamma * t)

t_env_scaled = t_env / t_dyn
t0_scaled = t_env_scaled.min()
t_fit = t_env_scaled - t0_scaled

popt, pcov = curve_fit(
    exp_decay,
    t_fit,
    d_env,
    p0=[d_env.max() - d_env.min(), 0.5, d_env.min()],
    bounds=([0, 0, 0], [np.inf, np.inf, np.inf]),
    maxfev=10000
)

A, gamma_scaled, d_floor = popt
perr = np.sqrt(np.diag(pcov))

tau_scaled = 1 / gamma_scaled
tau_scaled_err = perr[1] / gamma_scaled**2

gamma_phys = gamma_scaled / t_dyn
gamma_phys_err = perr[1] / t_dyn

tau_phys = 1 / gamma_phys
tau_phys_err = gamma_phys_err / gamma_phys**2

print(f"t_dyn          = {t_dyn:.4f} Gyr")
print(f"A              = {A:.4e} ± {perr[0]:.4e}")
print(f"gamma_scaled   = {gamma_scaled:.4e} ± {perr[1]:.4e} [1/t_dyn]")
print(f"tau_scaled     = {tau_scaled:.4e} ± {tau_scaled_err:.4e} [t_dyn]")
print(f"gamma_phys     = {gamma_phys:.4e} ± {gamma_phys_err:.4e} 1/Gyr")
print(f"tau_phys       = {tau_phys:.4e} ± {tau_phys_err:.4e} Gyr")
print(f"d_floor        = {d_floor:.4e} ± {perr[2]:.4e}")

# ============================================================
# Plot
# ============================================================

t_plot = np.linspace(t_start, t_env.max(), 500)
t_plot_scaled = t_plot / t_dyn

d_plot = exp_decay(t_plot_scaled - t0_scaled, *popt)

plt.figure(figsize=(9, 5))

plt.plot(
    time,
    dist,
    color="purple",
    label="sink-particle separation"
)

plt.scatter(
    t_env,
    d_env,
    color="black",
    s=25,
    label="separation maxima used for fit"
)

plt.plot(
    t_plot,
    d_plot,
    "r--",
    label=fr"fit: $\gamma t_{{dyn}}={gamma_scaled:.3f}$"
)

plt.axvline(
    t_start,
    color="gray",
    linestyle=":",
    label="fit start"
)

plt.axhline(
    d_min_resolved,
    color="gray",
    linestyle="-.",
    label="resolution cut"
)

plt.xlabel("Time [Gyr]")
plt.ylabel("Sink separation [kpc]")

plt.xlim(3.1, 10)
plt.ylim(-0.2, 6.5)

plt.legend()
plt.tight_layout()
plt.savefig("plots/distances/two_4comp_systems_D10,10_V0,2.png",dpi=300)
plt.show()
```

```python
import numpy as np
import os
import pandas as pd
from sklearn.cluster import DBSCAN

from scipy.interpolate import interp1d
from scipy.optimize import brentq

from scipy.optimize import newton
from multiprocessing import Process 

import warnings


G_KPC3_MSUN_GYR2 = 4.4986381856937695e-6
HBAR_TILDE = 7.842780148664155
Ncores = os.cpu_count()
AS = 0.0
EPS_KPC = 1e-10
M_BH = 4.3e6
def gpp(y, r, omega, a_s=AS):
    Phi, U, A, B = y
    rr = r if r > EPS_KPC else EPS_KPC
    Ubh = -G_KPC3_MSUN_GYR2 * M_BH / (HBAR_TILDE**2 * rr)
    Utot = U + Ubh

    if r == 0:
        dydr = [A, B, 2*(Utot + a_s*Phi**2 - omega)*Phi/3.0, 4*np.pi*Phi**2/3.0]
    else:
        dydr = [A, B, 2*(Utot + a_s*Phi**2 - omega)*Phi - (2.0/r)*A, 4*np.pi*Phi**2 - (2.0/r)*B]
    return dydr
    
def Sol(r, U0, omega):
    y0 = [1, U0, 0.0, 0.0]
    sol = odeint(gpp, y0, r, args=(omega,), mxstep=5000)
    return sol




def Phiinft(r,U0=-1.3473803,omega=-0.69831):
     return Sol(r,U0,omega)[:,0][m-1]
def recta(x,a,m):
    return x*m+a
def TotalMass(Malla,r):
    return 4*np.pi*np.sum((r[1]-r[0])*r**2*np.abs(Malla)**2)


Modos = np.loadtxt(r"Sols.txt")
# Modos = Modos[(Modos[:,0]>-10),:]
# Modos = Modos[(Modos[:,0]>-2.6)&(Modos[:,0]<-2.5),:]
# Modos = Modos[(Modos[:,1]<-1.5),:]
# Modos[:,1] = Modos[:,1]-Modos[:,0]
# Modos[:,1] = Modos[:,1]/100
```

```python
# %matplotlib qt
plt.scatter(Modos[:,0],Modos[:,1],s=0.1)

dbscan = DBSCAN(eps=0.1,n_jobs=Ncores)
dbscan.fit(Modos[:,1].reshape(-1, 1) )

labels = dbscan.labels_
dfdb = pd.DataFrame()
dfdb["U0"] = Modos[:,0]
dfdb["omega"] = Modos[:,1]
dfdb["rc"] = Modos[:,2]
dfdb["c"] = labels
dfdb = dfdb.sort_values(by=['omega','c']).reset_index(drop=True)
# Number of clusters in labels, ignoring noise if present.
n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
n_noise_ = list(labels).count(-1)

print("Estimated number of clusters: %d" % n_clusters_)
print("Estimated number of noise points: %d" % n_noise_)
unique_labels = set(labels)
core_samples_mask = np.zeros_like(labels, dtype=bool)
core_samples_mask[dbscan.core_sample_indices_] = True
plt.figure(figsize=(10,8))

colors = [plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels))]
for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = [0, 0, 0, 1]

    class_member_mask = labels == k

    xy = Modos[class_member_mask & core_samples_mask]
    plt.plot(
        xy[:, 0],
        xy[:, 1],
        "o",
        markerfacecolor=tuple(col),
        markeredgecolor=None,
        markersize=6,
    )

    xy = Modos[class_member_mask & ~core_samples_mask]
    plt.plot(
        xy[:, 0],
        xy[:, 1],
        "o",
        markerfacecolor=tuple(col),
        markeredgecolor="k",
        markersize=2,
    )

plt.title(f"Estimated number of clusters: {n_clusters_}")
plt.show()
```

```python
Clusters = sorted(dfdb["c"].drop_duplicates().to_numpy())

for c in Clusters:
    plt.figure()
    Vals = dfdb[dfdb["c"]==c][dfdb.columns[:3]].values
    if c != -1:
        Vals = Vals[np.random.choice(np.arange(0,len(Vals),1), np.min([int(len(Vals)/10+1),20]), replace=False)]
    for U0, omega, rn in Vals:
        try:
            r = np.linspace(EPS_KPC, rn, num=1000)
            Solution = Sol(r, U0, omega)
            plt.plot(r, Solution[:,0], label=f"{U0:.2e},{omega:.2e}")
        except Warning:
            pass
    plt.title(c)
    plt.show()
```

```python
def ShootTail(r, omega0, U0, omegalims, atol=1e-4, Plot=True):
    Solution = Sol(r, U0, omega0)
    while True:
        Shoot = [(omega0, Solution[-1,0])]

        Omega = np.random.uniform(omegalims[0], omegalims[1], 50)
        for omega in Omega:
            try:
                Solution = Sol(r, U0, omega)
                Shoot.append((omega, Solution[-1,0]))
            except Warning:
                pass
        Shoot = np.array(Shoot)

        if len(Shoot) > 1 and np.max(Shoot[:,1]) > 0 and np.min(Shoot[:,1]) < 0:
            omegalims = np.sort(
                Shoot[
                    (Shoot[:,1] == Shoot[Shoot[:,1] > 0, 1].min()) |
                    (Shoot[:,1] == Shoot[Shoot[:,1] < 0, 1].max()),
                    0
                ]
            )

            Omegas = np.linspace(omegalims[0], omegalims[1], 200)
            Lim = []
            for omega in Omegas:
                try:
                    Solution = Sol(r, U0, omega)
                    Lim.append(Solution[-1,0])
                except Warning:
                    Lim.append(np.nan)
            Lim = np.array(Lim)
            mask = np.isfinite(Lim)
            Omegas = Omegas[mask]
            Lim = Lim[mask]
            if len(Omegas) < 3:
                omegalims[0] -= 1e-2
                omegalims[1] += 1e-2
                continue

            Omegas_funct = interp1d(Omegas.astype(np.longdouble), Lim.astype(np.longdouble), kind="linear")
            try:
                new_omega = np.longdouble(brentq(Omegas_funct, Omegas.min(), Omegas.max()))
            except:
                new_omega = np.longdouble(0.5*(Omegas.min()+Omegas.max()))

            Solution = Sol(r, U0, new_omega)
            DifZero = np.abs(Solution[-1,0])

            if DifZero < atol:
                if Plot:
                    fig, ax = plt.subplots()
                    ax.plot(r, Solution[:,0], ls='--')
                    ax1 = plt.twinx()
                    ax1.semilogy(r, np.abs(Solution[:,0]), ls='--')
                    ax1.axhline(atol, ls='--')
                    plt.show()
                    plt.close("all")
                return new_omega, omegalims, Solution
            else:
                omega0 = float(new_omega)
                continue
        else:
            omegalims[0] -= 1e-2
            omegalims[1] += 1e-2
```

```python
U0,omega0 = u0origi,omegaorigi
U0_recta = np.linspace(-16,-0,5)
Omega_recta = recta(U0_recta,omega0-U0,1)#-AS
atol = 1e-6
Ushoot = []
Size = len(U0_recta)
Countet = 0

plt.figure(figsize=(7,5))
rr = np.where(r > EPS_KPC, r, EPS_KPC)
Ubh = -G_KPC3_MSUN_GYR2 * M_BH / (HBAR_TILDE**2 * rr)

for UU,omegaa in np.vstack([U0_recta,Omega_recta]).T:
    omega0,omegalims,Solution = ShootTail(r,omegaa,UU,[omegaa-atol,omegaa+atol],atol,False)
    if UU==U0_recta[0]:
        plt.plot(r,Solution[:,1],c=(Countet/Size,0.5,Countet/Size),label=r"$\phi_0 = {:.2f}$".format(UU))
    elif UU==U0_recta[-1]:
        plt.plot(r,Solution[:,1],c=(Countet/Size,0.5,Countet/Size),label=r"$\phi_0 = {:.2f}$".format(UU))
    else:
        plt.plot(r,Solution[:,1],c=(Countet/Size,0.5,Countet/Size))
    Masa = TotalMass(Solution[:,0],r)
    V_lim = -Masa/(r[-1])
    Ushoot.append((Solution[-1,1]-V_lim,omega0))
    Countet+=1
plt.plot(r[1:],-Masa/r[1:],ls='--',c='k',label='Expected Boundary')
plt.ylim(U0_recta.min(),0)

plt.xlabel(r"Distance $r$ [$\lambda_B$]",fontsize=16)
plt.ylabel(r"Solution $\phi$ [$v^2$]",fontsize=16)

legend = plt.legend(fontsize=12,loc="best", edgecolor=None)
legend.get_frame().set_alpha(0)
legend.get_frame().set_facecolor((0, 0, 1, 0.1))
plt.suptitle(r"Different Solutions for the potential $\phi$", fontsize=16)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)


plt.tight_layout()
plt.grid(True, which="both", ls="-")
plt.savefig("PotSolEigen.png",dpi=300)
plt.show()
Ushoot = np.array(Ushoot)
U_lim_funct = interp1d(U0_recta,Ushoot[:,0])
Omega_funct = interp1d(U0_recta,Ushoot[:,1])
```

```python
U = brentq(U_lim_funct, U0_recta.min(), U0_recta.max())
omega = float(Omega_funct(U))

# IMPORTANT: start r from EPS_KPC
r = np.linspace(EPS_KPC, float(r[-1]) if np.ndim(r) else 8.0, len(r))  # or define R,N explicitly below
# better explicit:
# R = 8.0
# N = 5*2048
# r = np.linspace(EPS_KPC, R, N)

omega, omegalims, Solution = ShootTail(r, omega, U, [omega-atol, omega+atol], atol, False)

# recompute final solution at the refined omega (keeps your “bounded” logic)
Solution = Sol(r, U0=U, omega=omega)

Msol = TotalMass(Solution[:,0], r)

# self boundary (consistent with Solution[:,1] being self-potential)
Uself_boundary = -Msol / r

# optional: total potential curve (self + BH) for visual sanity
Ubh = -G_KPC3_MSUN_GYR2 * M_BH / (HBAR_TILDE**2 * np.maximum(r, EPS_KPC))
Utot = Solution[:,1] + Ubh
Utot_boundary = -(Msol + M_BH) / r

fig = plt.figure(figsize=(6, 6))
gs = fig.add_gridspec(2, 1, hspace=0, wspace=0)
ax = gs.subplots(sharex='col', sharey='row')

ax[0].plot(r, Solution[:,0], label='Solution', c='tab:red')
ax[0].plot(r, Solution[:,2], label='Derivative', c='tab:green')

ax[1].plot(r, Solution[:,1], label='U self', c='tab:red')
ax[1].plot(r, Uself_boundary, label='Self boundary ~ -Msol/r', c='tab:blue', ls='--')

# optional (recommended): show total too
ax[1].plot(r, Utot, label='U total (self+BH)', c='tab:orange')
ax[1].plot(r, Utot_boundary, label='Total boundary ~ -(Msol+MBH)/r', c='tab:gray', ls='--')

ax[1].set_ylim(min(Solution[:,1].min(), Utot.min()) - 0.2, 0)

ax[1].set_ylabel(r"Potential")
ax[0].set_ylabel(r"Stationary solution $\varphi$")
ax[1].set_xlabel(r"Distance $r$")
ax[1].legend()
ax[0].legend()
ax[0].set_title(r"$M$ = {:.3f}".format(Msol))
ax[0].grid()
ax[1].grid()

plt.subplots_adjust(hspace=0.05)
plt.tight_layout()
plt.savefig("GroundState.png", dpi=300)
plt.show()
```
