```python
import os
import glob
import numpy as np
import pandas as pd
import astropy.units as u
import astropy.constants as const

# 1. Funkce pro načtení dat (vytaženo z tvé třídy)
def read_ground_state_data(file_path):
    try:
        with open(file_path, 'r') as f:
            lines = f.readlines()
        
        data_lines = [line for line in lines if not line.strip().startswith('#')]
        data = []
        for line in data_lines:
            values = line.strip().split()
            if len(values) >= 2:  # Potřebujeme jen r a phi
                r = float(values[0])
                phi = float(values[1])
                data.append([r, phi])
                
        return np.array(data, dtype=np.float32)
    except Exception as e:
        print(f"Chyba při čtení {file_path}: {e}")
        return None

# 2. Funkce pro výpočet fyzikální hmotnosti
def calculate_soliton_mass(file_path, G, h_bar_tilde):
    data = read_ground_state_data(file_path)
    if data is None or len(data) == 0:
        return np.nan
        
    r_values = data[:, 0]
    phi_values = data[:, 1]
    
    # Fyzikální přeškálování pole phi
    phi_physical = phi_values * (h_bar_tilde / np.sqrt(G))
    density = np.abs(phi_physical) ** 2  # rho(r) = |psi(r)|^2

    # Integrace M = ∫ rho(r) * 4πr² dr pomocí lichoběžníkového pravidla
    integrand = 4 * np.pi * r_values ** 2 * density
    mass = np.trapezoid(integrand, r_values)
    
    return mass

# 3. Parametry, které jsme projížděli
M_BH_list = [1e4, 5e4, 1e5, 5e5, 1e6, 5e6, 1e7, 5e7, 1e8, 2.5e8, 5e8, 7.5e8,1e9,2.5e9,5e9,7.5e9,1e10,2.5e10,5e10,1e11]
m_s_list = [1e-21, 1e-22, 1e-23]

# Tady budeme ukládat výsledky
results = []

print("Začínám zpracovávat soubory módů...")

for m_s_val in m_s_list:
    # A. Dynamický výpočet konstant pro daný axion
    m_s_eV = m_s_val * u.eV
    mass_s = (m_s_eV / const.c**2).to(u.Msun)
    G = const.G.to(u.kpc**3 / (u.Msun * u.Gyr**2)).value
    h_bar = const.hbar.to(u.Msun * u.kpc**2 / u.Gyr)
    h_bar_tilde = (h_bar / mass_s).to(u.kpc**2 / u.Gyr).value
    
    for M_BH in M_BH_list:
        # B. Rekonstrukce jména souboru (přesně jak ho ukládal tvůj kód)
        suffix = f"{M_BH:.0e}_{h_bar_tilde:.1e}".replace("+0", "").replace("+", "")
        fn = f"resources/solitons/MBH_{M_BH:.0e}_phi1/Modo_{suffix}_phi1.0.dat"
        
        if os.path.exists(fn):
            # C. Výpočet hmotnosti
            mass = calculate_soliton_mass(fn, G, h_bar_tilde)
            
            # D. Uložení do seznamu
            results.append({
                "m_s_eV": m_s_val,
                "M_bh": M_BH,
                "M_core": mass
            })
            print(f"Zpracováno: m_s = {m_s_val:.1e}, M_BH = {M_BH:.1e} -> M_core = {mass:.2e} Msun")
        else:
            print(f"Nenalezeno: {fn}")

# 4. Převod do Pandas tabulky a uložení
df_results = pd.DataFrame(results)

# Uložení do CSV, aby to bylo připravené na plotování
output_csv = "soliton_masses_scaling.csv"
df_results.to_csv(output_csv, index=False)

print(f"\nHotovo! Všechna data uložena do '{output_csv}'.")
display(df_results.head(10))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import os

def power_law_fit(M_bh, M_0, M_bh0, alpha):
    return M_0 / (1 + (M_bh / M_bh0)**alpha)

def log_power_law_fit(M_bh, M_0, M_bh0, alpha):
    return np.log10(power_law_fit(M_bh, M_0, M_bh0, alpha))

df = pd.read_csv("soliton_masses_scaling.csv")
m_s_list = sorted(df['m_s_eV'].unique(), reverse=True)

colors = ['black', 'royalblue', 'forestgreen', 'darkorange']
fit_colors = ['red', 'dodgerblue', 'lime', 'orange']
markers = ['o', 's', 'v', 'd']

plt.figure(figsize=(9, 7))

for i, ms in enumerate(m_s_list):
    df_ms = df[df['m_s_eV'] == ms].sort_values(by='M_bh')
    df_ms = df_ms.dropna(subset=['M_core'])

    # Remove non-positive values for log fit
    df_ms = df_ms[(df_ms['M_bh'] > 0) & (df_ms['M_core'] > 0)]

    if df_ms.empty:
        continue

    c_data = colors[i % len(colors)]
    c_fit = fit_colors[i % len(fit_colors)]
    mk = markers[i % len(markers)]

    xdata = df_ms['M_bh'].values
    ydata = df_ms['M_core'].values

    plt.plot(xdata, ydata, marker=mk, ms=8, color=c_data, ls='',
             label=fr'Data ($m_s = {ms:.0e}$ eV)')

    if len(df_ms) >= 3:
        p0 = [ydata.max(), np.median(xdata), 2.0]

        try:
            popt, pcov = curve_fit(
                log_power_law_fit,
                xdata,
                np.log10(ydata),
                p0=p0,
                bounds=([1e-20, 1e-20, 0.1], [np.inf, np.inf, 10.0]),
                maxfev=20000
            )

            x_fit = np.logspace(np.log10(xdata.min()), np.log10(xdata.max()), 300)
            y_fit = power_law_fit(x_fit, *popt)

            eq = fr'Fit: $M_0={popt[0]:.1e}, M_{{bh0}}={popt[1]:.1e}, \alpha={popt[2]:.2f}$'
            plt.plot(x_fit, y_fit, color=c_fit, ls='--', lw=1.5, label=eq)

        except Exception as e:
            print(f"Fit pro m_s={ms:.0e} selhal: {e}")

plt.xscale('log')
plt.yscale('log')

plt.xlabel(r'Black hole mass $M_{bh}\ [M_\odot]$', fontsize=14)
plt.ylabel(r'Core mass $M_{core}\ [M_\odot]$', fontsize=14)
plt.tick_params(axis='both', which='both', direction='out', length=5, labelsize=12)
plt.legend(fontsize=10, frameon=False, loc='lower left')
plt.tight_layout()

out_pdf = "plots/soliton_masses/Msol_scaling_with_ms.png"
os.makedirs(os.path.dirname(out_pdf), exist_ok=True)
plt.savefig(out_pdf, format='png', dpi=600, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import i0, i1, k0, k1

# ----------------------------
# constants / unit conversion
# ----------------------------
G = 4.4986381856937695e-6        # kpc^3 / (Msun * Gyr^2)
KM_S_TO_KPC_GYR = 1.0227121650537077
KPC_GYR_TO_KM_S = 1.0 / KM_S_TO_KPC_GYR

# ----------------------------
# load measured gas curve
# ----------------------------
rot_file = r"resources/data/two_system_orbit+collision_continuation2/rotational_velocity.dat"
df = pd.read_csv(rot_file)

t_target = 20.0
times = np.sort(df["time"].unique())
t_plot = times[np.argmin(np.abs(times - t_target))]

df_t = df[np.isclose(df["time"], t_plot)].copy()
dgas = df_t[df_t["component"] == "gas"].copy().sort_values("R")

print(f"Using measured gas rotation curve at t = {t_plot:.6f} Gyr")


MBH = 2*1e6       # Msun
Mgas = 2*2e7      # Msun
Mbulge = 2*3e7    # Msun
Msol = 2*3.3e8    # Msun


Rd_gas = 4.0      
Rcut_gas = 15.0   
a_bulge = 10.0   


Rhalf_sol = 3   
a_sol = Rhalf_sol / 1.305

print(f"Estimated soliton half-mass radius: {Rhalf_sol:.3f} kpc")
print(f"Plummer-equivalent soliton scale:    {a_sol:.3f} kpc")

# ----------------------------
# model functions
# ----------------------------
def vc_bh(R, M):
    R = np.asarray(R)
    return np.sqrt(G * M / R)


def vc_hernquist(R, M, a):
    R = np.asarray(R)
    return np.sqrt(G * M * R / (R + a)**2)


def vc_exp_disk(R, M, Rd):
    """
    Circular velocity of an infinitesimally thin exponential disk.

    v_c^2 = 4 pi G Sigma0 Rd y^2 [I0(y)K0(y) - I1(y)K1(y)]
    where y = R / (2 Rd), Sigma0 = M / (2 pi Rd^2).
    """
    R = np.asarray(R)
    y = R / (2.0 * Rd)
    Sigma0 = M / (2.0 * np.pi * Rd**2)

    bessel_term = i0(y) * k0(y) - i1(y) * k1(y)
    v2 = 4.0 * np.pi * G * Sigma0 * Rd * y**2 * bessel_term

    v2 = np.clip(v2, 0.0, None)
    return np.sqrt(v2)


def vc_plummer(R, M, a):
    R = np.asarray(R)
    return np.sqrt(G * M * R**2 / (R**2 + a**2)**1.5)


# ----------------------------
# calculation
# ----------------------------
R = np.linspace(0.1, 20.0, 400)

v_bh = vc_bh(R, MBH)
v_gas = vc_exp_disk(R, Mgas, Rd_gas)
v_bulge = vc_hernquist(R, Mbulge, a_bulge)
v_sol = vc_plummer(R, Msol, a_sol)

v_pred_no_dm = np.sqrt(
    v_bh**2
    + v_gas**2
    + v_bulge**2
)

v_pred_with_dm = np.sqrt(
    v_bh**2
    + v_gas**2
    + v_bulge**2
    + v_sol**2
)

# convert theory to km/s
v_bh_kms = v_bh * KPC_GYR_TO_KM_S
v_gas_kms = v_gas * KPC_GYR_TO_KM_S
v_bulge_kms = v_bulge * KPC_GYR_TO_KM_S
v_sol_kms = v_sol * KPC_GYR_TO_KM_S

v_pred_no_dm_kms = v_pred_no_dm * KPC_GYR_TO_KM_S
v_pred_with_dm_kms = v_pred_with_dm * KPC_GYR_TO_KM_S

# ----------------------------
# measured gas
# ----------------------------
Rm = dgas["R"].values
vm = dgas["vphi_mean"].values * KPC_GYR_TO_KM_S
sm = dgas["vphi_std"].values * KPC_GYR_TO_KM_S

vm_smooth = pd.Series(vm).rolling(3, center=True, min_periods=1).mean().values

# ----------------------------
# plot: total prediction vs measured gas
# ----------------------------
plt.figure(figsize=(8, 6))

plt.plot(
    Rm,
    vm_smooth,
    marker="o",
    ms=4,
    lw=1.8,
    color="tab:blue",
    label=f"measured gas rotation, t = {t_plot:.2f} Gyr",
)

plt.fill_between(
    Rm,
    vm_smooth - sm,
    vm_smooth + sm,
    color="tab:blue",
    alpha=0.15,
    label=r"gas $1\sigma$ scatter",
)

plt.plot(
    R,
    v_pred_no_dm_kms,
    "k--",
    lw=2.2,
    label="predicted: BH + gas disk + Hernquist bulge",
)

plt.plot(
    R,
    v_pred_with_dm_kms,
    "r-",
    lw=2.2,
    label="predicted: + ULDM soliton",
)

plt.xlabel("R [kpc]")
plt.ylabel(r"$v_{\rm circ}$ [km/s]")
plt.legend()
plt.tight_layout()
plt.show()

# ----------------------------
# optional diagnostic plot: individual components
# ----------------------------
plt.figure(figsize=(8, 6))

plt.plot(R, v_bh_kms, label="BH")
plt.plot(R, v_gas_kms, label="gas disk")
plt.plot(R, v_bulge_kms, label="Hernquist bulge")
plt.plot(R, v_sol_kms, label="ULDM soliton approx.")

plt.plot(R, v_pred_no_dm_kms, "k--", lw=2.2, label="total without ULDM")
plt.plot(R, v_pred_with_dm_kms, "r-", lw=2.2, label="total with ULDM")

plt.xlabel("R [kpc]")
plt.ylabel(r"$v_{\rm circ}$ [km/s]")
plt.legend()
plt.tight_layout()
plt.show()
```
