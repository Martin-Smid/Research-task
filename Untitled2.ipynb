```python
seeds = [1, 2, 3, 4, 5]
spins = [0, 1, 2, 3]

boundaries = [[-50, 50], [-50, 50], [-50, 50]]

def generate_random_position(boundary):
    low, high = boundary
    return [np.random.uniform(low, high), np.random.uniform(low, high), 0.0]

def is_far_enough(new_pos, existing_positions, min_dist):
    for pos in existing_positions:
        if np.linalg.norm(np.array(new_pos) - np.array(pos)) < min_dist:
            return False
    return True

for seed in seeds:
    for spin in spins:
        np.random.seed(seed)

        sim = Simulation_Class(
            dim=3,
            boundaries=boundaries,
            N=128,
            total_time=2.5,
            h=2.5e-3,
            order_of_evolution=2,
            use_gravity=True,
            static_potential=None,
            save_max_vals=True,
            self_int=False,
            use_sponge=False,
            m_s=m_scalar_eV,
        )

        waves = []
        positions = []
        min_separation = 8
        boundary = [-40, 40]

        for i in range(1):
            while True:
                means = generate_random_position(boundary)
                if is_far_enough(means, positions, min_separation):
                    positions.append(means)
                    break

            print(f"Seed {seed}, spin {spin}, wave {i+1} position: {means}")

            vlna = Wave_vector_class(
                packet_type="resources/solitons/GroundState(1).dat",
                means=means,
                st_deviations=[0.5, 0.5, 0.5],
                simulation=sim,
                mass=1,
                omega=1,
                momenta=[0.0, 0.0, 0.0],
                spin=spin,
                desired_soliton_mass=5.3090068e7,
            )
            sim.add_wave_vector(wave_vector=vlna)
            vlna.cleanup_wave_vector()
            del vlna

        sim.initialize_simulation()

        sim.evolution.scribe.snapshot_directory = f"resources/data/simulation_seed{seed}_spin{spin}"
        sim.evolution.scribe.max_vals_filename = f"resources/data/max_values_seed{seed}.csv"

        sim.evolve(save_every=250)
```
