```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
import re

from scipy.optimize import curve_fit
from scipy.stats import t

spin_colors = {
    0: "#1f77b4", # tlumenější modrá
    1: "#ff7f0e", # oranžová
    2: "#2ca02c", # zelená
    3: "#d62728", # červená
}
# ----------------------------
# Fit model
# ----------------------------
def saturation_func(t, a, b, c):
    return a + b * t / (1 + c * t)

def compute_lambda_inf(a, b, c):
    return a + b / c

def compute_tau99(a, b, c):
    lambda_inf = compute_lambda_inf(a, b, c)
    lambda_99 = 0.99 * lambda_inf
    A = lambda_99 - a
    denom = b - c * A
    if np.isclose(denom, 0):
        return np.nan
    tau_99 = A / denom
    return tau_99

# ----------------------------
# Helpers
# ----------------------------
def parse_spin_label(spin_label):
    """
    Converts e.g. 's=0' -> 0
    """
    if isinstance(spin_label, str):
        match = re.search(r"s\s*=\s*(-?\d+)", spin_label)
        if match:
            return int(match.group(1))
    return None

def read_max_values_file(file_name):
    """
    Reads one max_values_seedX.csv file with MultiIndex header.
    Returns a dict:
        {spin: pd.Series}
    where index = timestep, values = raw max density
    """
    df = pd.read_csv(file_name, comment="#", header=[0, 1], index_col=0)

    spin_data = {}
    for col in df.columns:
        N_label, spin_label = col
        spin = parse_spin_label(spin_label)
        if spin is None:
            continue

        series = df[col].astype(float).dropna()
        spin_data[spin] = series

    return spin_data

def normalize_lambda_rho(series):
    """
    λ_rho(t) = (rho(t)/rho(0))^(1/4)
    """
    rho0 = series.iloc[0]
    if rho0 == 0:
        raise ValueError("Initial max density is zero, cannot normalize.")
    return (series / rho0) ** 0.25

def fit_one_curve(x, y):
    """
    Fit one λ_rho(t) curve.
    Returns popt = [a,b,c]
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 5:
        raise RuntimeError("Not enough valid points for fit.")

    # Reasonable initial guess
    a0 = 1.0
    y_end = np.nanmedian(y[-min(5, len(y)):])
    c0 = 0.05
    b0 = max((y_end - a0) * c0, 1e-6)

    # bounds chosen to keep asymptotic saturating behaviour sensible
    p0 = [a0, b0, c0]
    bounds = ([0.5, 0.0, 1e-8], [10.0, 100.0, 100.0])

    popt, pcov = curve_fit(
        saturation_func, x, y,
        p0=p0,
        bounds=bounds,
        maxfev=20000
    )
    return popt, pcov

# ----------------------------
# Main analysis
# ----------------------------
def analyze_multi_seed_fits(
    pattern="resources/data/max_values_seed*.csv",
    spins=(0, 1, 2, 3),
    confidence=0.95,
    time_as_steps=True
):
    """
    Loads all seed files, groups curves by spin, fits each curve separately,
    then computes mean fitted curve + CI across seeds for each spin.

    Returns:
        results: dict keyed by spin
    """
    files = sorted(glob.glob(pattern))
    if not files:
        raise FileNotFoundError(f"No files found for pattern: {pattern}")

    all_by_spin = {spin: [] for spin in spins}

    # Load all files
    for file_name in files:
        spin_dict = read_max_values_file(file_name)

        for spin in spins:
            if spin not in spin_dict:
                print(f"Warning: spin {spin} missing in {os.path.basename(file_name)}")
                continue

            series = spin_dict[spin]
            lam = normalize_lambda_rho(series)

            x = lam.index.to_numpy(dtype=float)
            y = lam.to_numpy(dtype=float)

            try:
                popt, pcov = fit_one_curve(x, y)
            except Exception as e:
                print(f"Fit failed for file={os.path.basename(file_name)}, spin={spin}: {e}")
                continue

            all_by_spin[spin].append({
                "file": os.path.basename(file_name),
                "x": x,
                "y": y,
                "popt": popt,
                "pcov": pcov,
                "lambda_inf": compute_lambda_inf(*popt),
                "tau99": compute_tau99(*popt),
            })

    # Build ensemble results
    results = {}

    alpha = 1 - confidence

    for spin in spins:
        runs = all_by_spin[spin]
        if len(runs) == 0:
            print(f"Warning: no successful fits for spin {spin}")
            continue

        # Common x-grid: intersection by using shortest available x
        min_len = min(len(r["x"]) for r in runs)
        x_common = runs[0]["x"][:min_len]

        # Evaluate all fitted curves on common grid
        fitted_curves = []
        raw_curves = []
        lambda_infs = []
        tau99s = []

        for r in runs:
            x = r["x"][:min_len]
            y = r["y"][:min_len]
            popt = r["popt"]

            fitted = saturation_func(x_common, *popt)
            fitted_curves.append(fitted)
            raw_curves.append(y)
            lambda_infs.append(r["lambda_inf"])
            tau99s.append(r["tau99"])

        fitted_curves = np.array(fitted_curves)   # shape: (n_runs, n_t)
        raw_curves = np.array(raw_curves)
        lambda_infs = np.array(lambda_infs, dtype=float)
        tau99s = np.array(tau99s, dtype=float)

        n = fitted_curves.shape[0]
        mean_fit = np.mean(fitted_curves, axis=0)
        sd_fit = np.std(fitted_curves, axis=0, ddof=1) if n > 1 else np.zeros_like(mean_fit)
        sem_fit = sd_fit / np.sqrt(n) if n > 0 else np.zeros_like(mean_fit)

        if n > 1:
            tcrit = t.ppf(1 - alpha / 2, df=n - 1)
            ci_low = mean_fit - tcrit * sem_fit
            ci_high = mean_fit + tcrit * sem_fit
        else:
            ci_low = mean_fit.copy()
            ci_high = mean_fit.copy()

        results[spin] = {
            "runs": runs,
            "x": x_common,
            "mean_fit": mean_fit,
            "sd_fit": sd_fit,
            "ci_low": ci_low,
            "ci_high": ci_high,
            "lambda_inf_mean": np.nanmean(lambda_infs),
            "lambda_inf_sd": np.nanstd(lambda_infs, ddof=1) if n > 1 else 0.0,
            "tau99_mean": np.nanmean(tau99s),
            "tau99_sd": np.nanstd(tau99s, ddof=1) if n > 1 else 0.0,
            "lambda_infs": lambda_infs,
            "tau99s": tau99s,
            "n_runs": n,
            "raw_curves": raw_curves,
        }

    return results

# ----------------------------
# Plotting
# ----------------------------
def plot_ensemble_fits(results, show_raw=False, show_individual_fits=False):
    """
    Plot mean fitted curve + 95% CI band for each spin.
    """
    plt.figure(figsize=(10, 6))

    for spin in sorted(results.keys()):
        color = spin_colors.get(spin, None)
        res = results[spin]
        x = res["x"]
        mean_fit = res["mean_fit"]
        ci_low = res["ci_low"]
        ci_high = res["ci_high"]

        label = (
            f"spin = {spin}\n"
            f"$\\lambda_\\infty$ = {res['lambda_inf_mean']:.2f} ± {res['lambda_inf_sd']:.2f}"
        )

        if show_raw:
            for y in res["raw_curves"]:
                plt.plot(x, y, alpha=0.20, linewidth=1)

        if show_individual_fits:
            for run in res["runs"]:
                plt.plot(x, saturation_func(x, *run["popt"]), alpha=0.25, linewidth=1.0)

        plt.plot(x, mean_fit, linewidth=2.5, color=color, label=label)
        plt.fill_between(x, ci_low, ci_high, color=color, alpha=0.1)

    plt.xlabel("Time step [t]", fontsize=16)
    plt.ylabel(r"$\lambda_\rho~[t]$", fontsize=16)
    plt.xticks(fontsize=13)
    plt.yticks(fontsize=13)
    plt.legend(fontsize=12)
    plt.tight_layout()
    plt.savefig(r"plots/solitonic_growth_mean.png",dpi=300)
    plt.show()

# ----------------------------
# Optional summary table
# ----------------------------
def make_summary_table(results):
    rows = []
    for spin in sorted(results.keys()):
        res = results[spin]
        rows.append({
            "spin": spin,
            "n_runs": res["n_runs"],
            "lambda_inf_mean": res["lambda_inf_mean"],
            "lambda_inf_sd": res["lambda_inf_sd"],
            "tau99_mean": res["tau99_mean"],
            "tau99_sd": res["tau99_sd"],
        })
    return pd.DataFrame(rows)

# ----------------------------
# Run
# ----------------------------
results = analyze_multi_seed_fits(
    pattern="resources/data/max_values_seed*.csv",
    spins=(0, 1, 2, 3),
    confidence=0.95
)

plot_ensemble_fits(
    results,
    show_raw=False,            # True if you want faint raw curves in background
    show_individual_fits=False  # useful to see seed-to-seed variation
)

summary_df = make_summary_table(results)
print(summary_df)
```

```python
import os
import shutil
import numpy as np

m_scalar_eV = 1e-22
HBAR_TILDE = 19.60695038367394

seeds = [1, 2, 3, 4, 5]
uldm_options = [False, True]

boundaries = [[-25, 25], [-25, 25], [-25, 25]]

for uldm_on in uldm_options:
    for seed in seeds:
        print("=" * 80)
        print(f"Running seed={seed}, uldm_on={uldm_on}")
        print("=" * 80)

        np.random.seed(seed)

        sim = Simulation_Class(
            dim=3,
            boundaries=boundaries,
            N=128,
            total_time=2.0,
            h=1e-3,
            order_of_evolution=2,
            use_gravity=True,
            static_potential=None,
            save_max_vals=True,
            self_int=False,
            use_sponge=True,
            m_s=m_scalar_eV,
            sink_formation={
                "source": "both",
                "consecutive_steps": 5,
                "check_interval": 5,

                "enable_gas_sinks": True,
                "gas_mode": "truelove",
                "gas_NJ": 4,
                "gas_cs_floor": 0.01,
                "gas_consecutive_steps": 5,
                "gas_check_interval": 5,
                "gas_r_acc_cells": 3,

                "merge_r_cells": 2,
                "merge_interval": 10,
            }
        )

        # ------------------------------------------------------------
        # Gas initialization
        # ------------------------------------------------------------
        N = sim.N
        x, y, z = sim.grids
        x = np.array(x)
        y = np.array(y)
        z = np.array(z)

        rho0 = 1.0 / (sim.dV * (N ** sim.dim))
        A = 200.0 * rho0
        sigma = 5.0 * min(sim.dx)

        r2 = x**2 + y**2 + z**2
        rho = rho0 + A * np.exp(-0.5 * r2 / sigma**2)

        # Small seed-dependent perturbation.
        # This makes the different seeds actually independent realizations,
        # without strongly changing the global initial condition.
        perturb_amp = 1e-3
        rho *= 1.0 + perturb_amp * np.random.normal(size=rho.shape)

        rho = np.maximum(rho, 1e-12)

        Mgas_target = 6e5
        M_current = rho.sum() * sim.dV
        rho *= Mgas_target / M_current

        gas = NBodyGas(
            simulation=sim,
            rho=rho,
            cs=0.05,
            gamma=5/3,
            tcool=0.5,
            e_floor=0.0,
            cfl=0.4,
            max_substeps=200,
        )

        sim.add_baryons(gas)

        # ------------------------------------------------------------
        # Optional ULDM soliton
        # ------------------------------------------------------------
        if uldm_on:
            wave_vector = Wave_vector_class(
                packet_type="resources/solitons/GroundState(1).dat",
                means=[0.0, 0.0, 0.0],
                st_deviations=[1.2, 1.2, 1.2],
                simulation=sim,
                mass=1,
                omega=1,
                momenta=[0.0, 0.0, 0.0],
                spin=0,
                desired_soliton_mass=5e7,
            )

            sim.add_wave_vector(wave_vector)

            try:
                wave_vector.cleanup_wave_vector()
            except Exception:
                pass

            del wave_vector

        # ------------------------------------------------------------
        # Run simulation
        # ------------------------------------------------------------
        sim.initialize_simulation()
        sim.evolve(save_every=250)

        # ------------------------------------------------------------
        # Rename output directory
        # ------------------------------------------------------------
        tag = "uldm_on" if uldm_on else "uldm_off"
        desired_dir = f"resources/data/gas_uldm_seed{seed}_{tag}"

        source_dir = sim.evolution.scribe.snapshot_directory

        if os.path.exists(desired_dir):
            shutil.rmtree(desired_dir)

        shutil.move(source_dir, desired_dir)

        print(f"Saved run to: {desired_dir}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Settings
# ------------------------------------------------------------
base_dir = "resources/data"
seeds = [1, 2, 3, 4, 5]

groups = {
    "Gas only": {
        "tag": "uldm_off",
        "linestyle": "-",
    },
    "Gas + ULDM": {
        "tag": "uldm_on",
        "linestyle": "-",
    },
}

out_dir = os.path.join(base_dir, "gas_uldm_comparison_plots")
os.makedirs(out_dir, exist_ok=True)

# 95% CI for n=5: t_{0.975, df=4} ≈ 2.776
T_95_N5 = 2.776


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def load_run(seed, tag):
    run_dir = os.path.join(base_dir, f"gas_uldm_seed{seed}_{tag}")
    csv_path = os.path.join(run_dir, "sink_debug_log.csv")

    if not os.path.exists(csv_path):
        print(f"[warning] Missing file: {csv_path}")
        return None

    df = pd.read_csv(csv_path)
    df = df.sort_values("time").reset_index(drop=True)

    df["M_formed_cumulative"] = df["M_formed_step"].cumsum()
    df["M_accreted_cumulative"] = df["M_accreted_step"].cumsum()
    df["M_total_cumulative"] = df["M_formed_cumulative"] + df["M_accreted_cumulative"]

    if "M_total_gas_plus_sink" in df.columns:
        M0 = df["M_total_gas_plus_sink"].iloc[0]
        df["mass_conservation_rel"] = (df["M_total_gas_plus_sink"] - M0) / M0

    return df


def interpolate_group(dfs, column, n_grid=500):
    # Common time interval shared by all runs
    t_min = max(df["time"].min() for df in dfs)
    t_max = min(df["time"].max() for df in dfs)
    t_grid = np.linspace(t_min, t_max, n_grid)

    y = []
    for df in dfs:
        y_interp = np.interp(t_grid, df["time"].values, df[column].values)
        y.append(y_interp)

    y = np.asarray(y)
    mean = np.mean(y, axis=0)
    sd = np.std(y, axis=0, ddof=1)
    sem = sd / np.sqrt(y.shape[0])
    ci95 = T_95_N5 * sem

    return t_grid, mean, ci95, sd


def plot_mean_ci(column, ylabel, filename, title=None, yscale=None):
    plt.figure(figsize=(8, 5))

    for label, cfg in groups.items():
        dfs = []
        for seed in seeds:
            df = load_run(seed, cfg["tag"])
            if df is not None and column in df.columns:
                dfs.append(df)

        if len(dfs) == 0:
            print(f"[warning] No valid data for {label}, column {column}")
            continue

        t, mean, ci95, sd = interpolate_group(dfs, column)

        line, = plt.plot(
            t,
            mean,
            linestyle=cfg["linestyle"],
            linewidth=2,
            label=f"{label} mean",
        )

        color = line.get_color()
        plt.fill_between(
            t,
            mean - ci95,
            mean + ci95,
            color=color,
            alpha=0.20,
            linewidth=0,
            label=f"{label} 95% CI",
        )

    plt.xlabel("Time [Gyr]")
    plt.ylabel(ylabel)

    if title is not None:
        plt.title(title)

    if yscale is not None:
        plt.yscale(yscale)

    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, filename), dpi=300)
    plt.show()


# ------------------------------------------------------------
# Main plots
# ------------------------------------------------------------

# 1) Most important: cumulative accreted gas mass after sink formation
plot_mean_ci(
    column="M_accreted_cumulative",
    ylabel=r"Cumulative accreted gas mass [$M_\odot$]",
    filename="01_mean_cumulative_accreted_mass_CI.png",
)

# 2) Total sink-related mass: formed seed + subsequent accretion
plot_mean_ci(
    column="M_total_cumulative",
    ylabel=r"Cumulative sink-related gas mass [$M_\odot$]",
    filename="02_mean_total_formed_plus_accreted_CI.png",
)

# 3) Threshold diagnostic
plot_mean_ci(
    column="rho_max_over_thr",
    ylabel=r"$\rho_{\rm max}/\rho_{\rm thr}$",
    filename="03_mean_rho_max_over_threshold_CI.png",
)

# 4) Optional: mass conservation check
plot_mean_ci(
    column="mass_conservation_rel",
    ylabel=r"$(M_{\rm gas}+M_{\rm sink}-M_0)/M_0$",
    filename="04_mean_mass_conservation_CI.png",
)

print(f"Plots saved to: {out_dir}")
```

```python
import os
import numpy as np
import pandas as pd

base_dir = "resources/data"
seeds = [1, 2, 3, 4, 5]

groups = {
    "Gas only": "uldm_off",
    "Gas + ULDM": "uldm_on",
}

target_time = 2.0

rows = []

for label, tag in groups.items():
    values = []

    for seed in seeds:
        run_dir = os.path.join(base_dir, f"gas_uldm_seed{seed}_{tag}")
        csv_path = os.path.join(run_dir, "sink_debug_log.csv")

        df = pd.read_csv(csv_path).sort_values("time").reset_index(drop=True)
        df["M_accreted_cumulative"] = df["M_accreted_step"].cumsum()

        # value closest to 2 Gyr
        idx = (df["time"] - target_time).abs().idxmin()
        M_acc = df.loc[idx, "M_accreted_cumulative"]
        t_used = df.loc[idx, "time"]

        values.append(M_acc)

        rows.append({
            "group": label,
            "seed": seed,
            "time_used": t_used,
            "M_acc_2Gyr": M_acc,
        })

    values = np.array(values)

    print(f"{label}:")
    print(f"  M_acc({target_time:g} Gyr) = {values.mean():.3f} ± {values.std(ddof=1):.3f} Msun")
    print(f"  individual seeds = {values}")
    print()

summary_df = pd.DataFrame(rows)
summary_df
```

```python
gas_only = summary_df[summary_df["group"] == "Gas only"]["M_acc_2Gyr"].values
gas_uldm = summary_df[summary_df["group"] == "Gas + ULDM"]["M_acc_2Gyr"].values

enhancement_ratio = gas_uldm / gas_only
relative_increase = (gas_uldm - gas_only) / gas_only

print(f"Enhancement ratio:")
print(f"  {enhancement_ratio.mean():.3f} ± {enhancement_ratio.std(ddof=1):.3f}")

print(f"Relative increase:")
print(f"  {100 * relative_increase.mean():.1f} ± {100 * relative_increase.std(ddof=1):.1f} %")
```
